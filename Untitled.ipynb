```python
# Cell 1
import pandas as pd
import numpy as np
from sklearn.linear_model import SGDRegressor
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt
```

```python
# Cell 2
# Load the dataset
data = pd.read_csv('house_prices.csv')

# Display the first few rows and basic info
print(data.head())
print("\nDataset Info:")
print(data.info())
```

```python
# Cell 3
# Drop non-predictive columns
data = data.drop(columns=['Order', 'PID'])

# Fill numeric NA with mean (e.g., Lot Frontage)
data = data.fillna(data.mean(numeric_only=True))

# Encode categorical variables
data = pd.get_dummies(data, drop_first=True)

# Display the processed data info
print(data.info())
```

```python
# Cell 4
# Features and target
X = data.drop(columns=['SalePrice'])
y = data['SalePrice']

# Scale the features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Split into training and test sets
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)

print("X_train shape:", X_train.shape)
print("y_train shape:", y_train.shape)
```

```python
# Cell 5 
from sklearn.linear_model import LinearRegression

# Initialize and train the model
model = LinearRegression()
model.fit(X_train, y_train)

# Print the scores
print(f"Training Score: {model.score(X_train, y_train):.4f}")
print(f"Test Score: {model.score(X_test, y_test):.4f}")
```

```python
# Cell 6
# Predict on test set
y_pred = model.predict(X_test)

# Plot
import matplotlib.pyplot as plt
plt.figure(figsize=(10, 6))
plt.scatter(y_test, y_pred, color='blue', alpha=0.5)
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'r--', lw=2)
plt.xlabel('Actual Price')
plt.ylabel('Predicted Price')
plt.title('Actual vs Predicted House Prices')
plt.tight_layout()
plt.show()
```

```python
# Cell 7
from sklearn.metrics import mean_squared_error

# Calculate RMSE
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
print(f"Root Mean Squared Error: {rmse:.2f}")

# Next steps suggestion
print("\nNext Steps: The RMSE shows the average error in dollars. If it’s high (e.g., > $50,000), consider feature selection, removing outliers, or trying a model like Random Forest. Let me know if you want to explore these!")
```

```python
# Cell 8 (Revised)
from sklearn.ensemble import RandomForestRegressor

# Train a Random Forest to get feature importance
rf_model = RandomForestRegressor(n_estimators=100, random_state=42)
rf_model.fit(X_train, y_train)

# Get feature importance
feature_importance = pd.DataFrame({
    'feature': data.drop(columns=['SalePrice']).columns,
    'importance': rf_model.feature_importances_
})
feature_importance = feature_importance.sort_values('importance', ascending=False)

# Select top 100 features
top_features = feature_importance.head(100)['feature'].values
X_train_selected = X_train[:, [list(data.drop(columns=['SalePrice']).columns).index(f) for f in top_features]]
X_test_selected = X_test[:, [list(data.drop(columns=['SalePrice']).columns).index(f) for f in top_features]]

# Train LinearRegression with selected features
model_selected = LinearRegression()
model_selected.fit(X_train_selected, y_train)

# Print new scores
print(f"Training Score (Top 100 Features): {model_selected.score(X_train_selected, y_train):.4f}")
print(f"Test Score (Top 100 Features): {model_selected.score(X_test_selected, y_test):.4f}")

# Calculate new RMSE
y_pred_selected = model_selected.predict(X_test_selected)
rmse_selected = np.sqrt(mean_squared_error(y_test, y_pred_selected))
print(f"Root Mean Squared Error (Top 100 Features): {rmse_selected:.2f}")
```

```python
# Cell 9
from sklearn.ensemble import RandomForestRegressor

# Train Random Forest with the selected 100 features
rf_final_model = RandomForestRegressor(n_estimators=100, random_state=42)
rf_final_model.fit(X_train_selected, y_train)

# Print scores
print(f"Training Score (Random Forest, 100 Features): {rf_final_model.score(X_train_selected, y_train):.4f}")
print(f"Test Score (Random Forest, 100 Features): {rf_final_model.score(X_test_selected, y_test):.4f}")

# Calculate RMSE
y_pred_rf = rf_final_model.predict(X_test_selected)
rmse_rf = np.sqrt(mean_squared_error(y_test, y_pred_rf))
print(f"Root Mean Squared Error (Random Forest, 100 Features): {rmse_rf:.2f}")
```

```python
# Cell 10
# Plot
plt.figure(figsize=(10, 6))
plt.scatter(y_test, y_pred_rf, color='blue', alpha=0.5)
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'r--', lw=2)
plt.xlabel('Actual Price')
plt.ylabel('Predicted Price')
plt.title('Actual vs Predicted House Prices (Random Forest)')
plt.tight_layout()
plt.show()
```

```python
# Cell 8
import joblib

# Save the Random Forest model
joblib.dump(rf_final_model, 'random_forest_house_price_model.pkl')

# Save the scaler (needed for future predictions)
joblib.dump(scaler, 'scaler.pkl')

print("Model and scaler saved successfully!")
```
